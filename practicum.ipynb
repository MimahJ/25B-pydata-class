1. Calculate the average enrolment per province. Which province has the highest average enrolment of students?

```python
import pandas as pd

df = pd.read_csv("kenya_primary_schools.csv")
```

```python

print(df.columns)  

avg_enrolment = df.groupby('Province')['TotalEnrol'].mean()

highest_province = avg_enrolment.idxmax()
highest_value = avg_enrolment.max()

print("Average Enrolment per Province:\n", avg_enrolment)
print(f"\nProvince with the highest average enrolment: {highest_province} ({highest_value})")
```

2. Which schools in Coast province are in most need of classrooms?

```python
import pandas as pd

# Filter only schools in Coast province
coast_schools = df[df["Province"] == "COAST"].copy()

# Ensure required columns are numeric
coast_schools["TotalEnrol"] = pd.to_numeric(coast_schools["TotalEnrol"], errors="coerce")
coast_schools["ClassrmRat"] = pd.to_numeric(coast_schools["ClassrmRat"], errors="coerce")
coast_schools["No_Classrm"] = pd.to_numeric(coast_schools["No_Classrm"], errors="coerce")

# Calculate classroom shortage
coast_schools["Classroom_Shortage"] = (coast_schools["TotalEnrol"] / coast_schools["ClassrmRat"]) - coast_schools["No_Classrm"]

# Fill NaN values with 0 before rounding
coast_schools["Classroom_Shortage"] = coast_schools["Classroom_Shortage"].fillna(0).apply(lambda x: round(x))

# Sort schools by highest classroom shortage
schools_in_need = coast_schools.sort_values(by="Classroom_Shortage", ascending=False)

# Display top 10 schools in need
print(schools_in_need[["School_Name", "Classroom_Shortage"]].head(10))
```

```python
# Filter dataset for Rift Valley Province
rift_schools = df[df["Province"] == "RIFT VALLEY"]

# Calculate total teachers per school
rift_schools["No_Teachers"] = rift_schools["GO_KTSC_M"] + rift_schools["GOK_TSC_F"]

# Assume 1 toilet per 5 teachers
rift_schools["Required_TeachersTo"] = (rift_schools["No_Teachers"] / 5).apply(lambda x: round(x))

# Calculate shortage of teacher toilets
rift_schools["Teacher_Toilet_Shortage"] = rift_schools["Required_TeachersTo"] - rift_schools["TeachersTo"]

# Sort schools by highest shortage
schools_most_in_need = rift_schools.sort_values(by="Teacher_Toilet_Shortage", ascending=False)

# Display the top 10 schools most in need
schools_most_in_need[["Name_of_Sc", "GO_KTSC_M", "GOK_TSC_F", "No_Teachers", "TeachersTo", "Required_TeachersTo", "Teacher_Toilet_Shortage"]].head(10)
```

4. Which schools in Central are in most need of sanitation facilities for students?

```python
import pandas as pd

# Filter only schools in Central Province
central_schools = df[df["Province"] == "CENTRAL"].copy()

# Ensure relevant columns are numeric
columns_to_convert = ["TotalEnrol", "BoysToilet", "GirlsToilet"]
for col in columns_to_convert:
    central_schools[col] = pd.to_numeric(central_schools[col], errors="coerce")

# Replace zero or NaN values in toilets with a small value (to avoid division by zero)
central_schools["BoysToilet"] = central_schools["BoysToilet"].replace(0, 1).fillna(1)
central_schools["GirlsToilet"] = central_schools["GirlsToilet"].replace(0, 1).fillna(1)

# Calculate Student-to-Toilet Ratio
central_schools["Toilet_Ratio"] = central_schools["TotalEnrol"] / (central_schools["BoysToilet"] + central_schools["GirlsToilet"])

# Sort schools by highest Student-to-Toilet Ratio (higher ratio = greater need)
schools_in_need = central_schools.sort_values(by="Toilet_Ratio", ascending=False)

# Display top 10 schools in need of toilets
print(schools_in_need[["Name_of_Sc", "TotalEnrol", "BoysToilet", "GirlsToilet", "Toilet_Ratio"]].head(10))
```

5. Which district in Nairobi County has the highest number of schools? Hint: Can you try using count() and size(), what difference do you notice (if any)

```python
import pandas as pd

# Filter only schools in Nairobi County
nairobi_schools = df[df["Province"] == "NAIROBI"].copy()

# Group by District and count the number of schools
district_school_counts = nairobi_schools.groupby("District")["Name_of_Sc"].count()

# Using size() to compare with count()
district_school_sizes = nairobi_schools.groupby("District").size()

# Find the district with the highest number of schools
most_schools_district = district_school_counts.idxmax()
most_schools_count = district_school_counts.max()

# Display results
print("Number of schools per district using count():\n", district_school_counts)
print("\nNumber of schools per district using size():\n", district_school_sizes)

print(f"\nDistrict with the highest number of schools: {most_schools_district} ({most_schools_count} schools)")
```

7. Add another column called ‘Total_school_staff’ and calculate the total number of school staff (teaching and non-teaching) in each school in Nairobi.

```python
import pandas as pd

# Filter only schools in Nairobi County
nairobi_schools = df[df["Province"] == "NAIROBI"].copy()

# Handle missing values (fill NaN with 0 to avoid calculation errors)
nairobi_schools[["GO_KTSC_M", "GOK_TSC_F", "NoTeaching", "NonTeachin"]] = nairobi_schools[
    ["GO_KTSC_M", "GOK_TSC_F", "NoTeaching", "NonTeachin"]
].fillna(0)

# Calculate total teaching staff
nairobi_schools["Total_Teachers"] = nairobi_schools["GO_KTSC_M"] + nairobi_schools["GOK_TSC_F"]

# Calculate total non-teaching staff
nairobi_schools["Total_NonTeaching"] = nairobi_schools["NoTeaching"] + nairobi_schools["NonTeachin"]

# Calculate total school staff
nairobi_schools["Total_school_staff"] = nairobi_schools["Total_Teachers"] + nairobi_schools["Total_NonTeaching"]

# Display results
print(nairobi_schools[["Total_Teachers", "Total_NonTeaching", "Total_school_staff"]].head())
```

8. Create a pivot table that shows the average number of classrooms (No_Classrm) and enrolment (TotalEnrol) by Province and District.

```python
import pandas as pd

# Handle missing values (fill NaN with 0 to avoid errors)
df[["No_Classrm", "TotalEnrol"]] = df[["No_Classrm", "TotalEnrol"]].fillna(0)

# Create the pivot table
pivot_table = df.pivot_table(
    values=["No_Classrm", "TotalEnrol"],  # Columns to calculate averages for
    index=["Province", "District"],       # Grouping by Province & District
    aggfunc="mean"                         # Calculate average
)

# Display results
print(pivot_table.head())
```
