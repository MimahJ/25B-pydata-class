### Introduction:

This time we will create our own dataset with fictional numbers to describe a house market. As we are going to create random data don't try to reason of the numbers.

### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Create 3 differents Series, each of length 100, as follows: 
1. The first a random number from 1 to 4 
2. The second a random number from 1 to 3
3. The third a random number from 10,000 to 30,000

```python
import numpy as np
series1 = pd.Series(np.random.randint(1, 5, size=100))

series2 = pd.Series(np.random.randint(1, 4, size=100))

series3 = pd.Series(np.random.randint(10000, 30001, size=100))

print("Series 1:")
print(series1.head())

print("\nSeries 2:")
print(series2.head())

print("\nSeries 3:")
print(series3.head())
```

### Step 3. Let's create a DataFrame by joinning the Series by column

```python
# Concatenate the Series into a DataFrame and assign column names
df_series = pd.concat([series1, series2, series3], axis=1)
df_series.columns = ['col1', 'col2', 'col3']

print(df_series.head())
```

### Step 4. Change the name of the columns to bedrs, bathrs, price_sqr_meter

```python
df_series.columns = ['bedrs', 'bathrs', 'price_sqr_meter']
print(df_series.head())
```

### Step 5. Create a one column DataFrame with the values of the 3 Series and assign it to 'bigcolumn'

```python
bigcolumn = pd.concat([series1, series2, series3], ignore_index=True).to_frame(name='bigcolumn')
print(bigcolumn.head(10))
print("Total rows:", len(bigcolumn))
```
