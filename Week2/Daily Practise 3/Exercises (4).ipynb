### Step 8. Group the year by decades and sum the values

#### Pay attention to the Population column number, summing this column is a mistake

```python
crime['Year'] = pd.to_datetime(crime['Year'], format='%Y')

crime.set_index('Year', inplace=True)

agg_dict = {col: 'sum' for col in crime.select_dtypes(include=['int64', 'float64']).columns}
if 'Population' in crime.columns:
    agg_dict['Population'] = 'first'

crime_decade = crime.groupby(pd.Grouper(freq='10Y')).agg(agg_dict)

print(crime_decade)
```

### Step 9. What is the most dangerous decade to live in the US?

```python
agg_dict = {col: 'sum' for col in crime.select_dtypes(include=['int64', 'float64']).columns}

if 'Population' in crime.columns:
    agg_dict['Population'] = 'first'

crime_decade = crime.groupby(pd.Grouper(freq='10Y')).agg(agg_dict)

crime_columns = [col for col in crime_decade.columns if col != 'Population']
crime_decade['Crime_total'] = crime_decade[crime_columns].sum(axis=1)

most_dangerous_decade = crime_decade['Crime_total'].idxmax()
print("The most dangerous decade is:", most_dangerous_decade)
print(crime_decade)
```
