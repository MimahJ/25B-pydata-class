# US - Baby Names

### Introduction:

We are going to use a subset of [US Baby Names](https://www.kaggle.com/kaggle/us-baby-names) from Kaggle.  
In the file it will be names from 2004 until 2014


### Step 1. Import the necessary libraries

```python
import pandas as pd
```

### Step 2. Import the dataset from this [address](https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv). 

### Step 3. Assign it to a variable called baby_names.

```python

url = "https://raw.githubusercontent.com/guipsamora/pandas_exercises/master/06_Stats/US_Baby_Names/US_Baby_Names_right.csv"

baby_names = pd.read_csv(url)
```

### Step 4. See the first 10 entries

```python
print(baby_names.head(10))
```

### Step 5. Delete the column 'Unnamed: 0' and 'Id'

```python
baby_names.drop(['Unnamed: 0', 'Id'], axis=1, inplace=True)

print(baby_names.head())
```

### Step 6. Is there more male or female names in the dataset?

```python

print(baby_names['Gender'].value_counts())

print(baby_names.groupby('Gender')['Name'].nunique())
```

### Step 7. Group the dataset by name and assign to names

```python
names = baby_names.groupby('Name')

print(list(names.groups.keys())[:10])
```

### Step 8. How many different names exist in the dataset?

```python

print("Different names in the dataset:", names.ngroups)
```

### Step 9. What is the name with most occurrences?

```python
name_counts = baby_names['Name'].value_counts()

most_common_name = name_counts.idxmax()
occurrence = name_counts.max()

print("Name with most occurrences:", most_common_name, "with", occurrence, "occurrences")
```

### Step 10. How many different names have the least occurrences?

```python
name_counts = baby_names['Name'].value_counts()

min_occurrence = name_counts.min()

least_occurrence_count = (name_counts == min_occurrence).sum()

print("Different names with the least occurrences:", least_occurrence_count)
```

### Step 11. What is the median name occurrence?

```python

name_counts = baby_names['Name'].value_counts()

median_occurrence = name_counts.median()

print("Median name occurrence:", median_occurrence)
```

### Step 12. What is the standard deviation of names?

```python
name_counts = baby_names['Name'].value_counts()

std_occurrence = name_counts.std()

print("Standard deviation of name occurrences:", std_occurrence)
```

### Step 13. Get a summary with the mean, min, max, std and quartiles.

```python
name_counts = baby_names['Name'].value_counts()

std_occurrence = name_counts.std()

print("Standard deviation of name occurrences:", std_occurrence)
```
