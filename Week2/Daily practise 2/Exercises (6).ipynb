### Step 3. Assign it to a variable called regiment.
#### Don't forget to name each column

```python

regiment = pd.DataFrame({
    'Regiment': ['Nighthawks', 'Nighthawks', 'Nighthawks', 'Nighthawks', 
                 'Dragoons', 'Dragoons', 'Dragoons', 'Dragoons', 
                 'Scouts', 'Scouts', 'Scouts', 'Scouts'],
    'Company': ['1st', '1st', '2nd', '2nd', 
                '1st', '1st', '2nd', '2nd',
                '1st', '1st', '2nd', '2nd'],
    'Name': ['Miller', 'Jacobson', 'Ali', 'Milner', 
             'Cooze', 'Jacon', 'Ryaner', 'Sone', 
             'Sloan', 'Piger', 'Riani', 'Ali'],
    'PreTestScore': [4, 24, 31, 2, 3, 4, 24, 31, 2, 3, 2, 3],
    'PostTestScore': [25, 94, 57, 62, 70, 25, 94, 57, 62, 70, 62, 70]
})

print(regiment)
```

### Step 4. What is the mean preTestScore from the regiment Nighthawks?  

```python
mean_pretest_nighthawks = regiment[regiment['Regiment'] == 'Nighthawks']['PreTestScore'].mean()

print(mean_pretest_nighthawks)
```

### Step 5. Present general statistics by company

```python
company_stats = regiment.groupby('Company').describe()
print(company_stats)

```

### Step 6. What is the mean of each company's preTestScore?

```python
company_mean_preTestScore = regiment.groupby('Company')['PreTestScore'].mean()
print(company_mean_preTestScore)
```

### Step 7. Present the mean preTestScores grouped by regiment and company

```python
mean_preTestScore = regiment.groupby(['Regiment', 'Company'])['PreTestScore'].mean()
print(mean_preTestScore)
```

### Step 8. Present the mean preTestScores grouped by regiment and company without heirarchical indexing

```python
mean_preTestScore = (
    regiment.groupby(['Regiment', 'Company'], as_index=False)['PreTestScore']
    .mean()
)

print(mean_preTestScore)
```

### Step 9. Group the entire dataframe by regiment and company

```python
grouped_df = regiment.groupby(['Regiment', 'Company']).mean(numeric_only=True)
print(grouped_df)
```

### Step 10. What is the number of observations in each regiment and company

```python
observations_count = regiment.groupby(['Regiment', 'Company']).size()
print(observations_count)
```

### Step 11. Iterate over a group and print the name and the whole data from the regiment

```python
grouped = regiment.groupby('Regiment')

for name, group in grouped:
    print(f"Regiment: {name}")
    print(group)
    print("-" * 40)  # Separator for readability
```
