### Step 4. How many products cost more than $10.00?

```python
import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")

# Step 1: Display unique item prices to check for formatting issues
print("Unique item prices (before cleaning):")
print(chipo["item_price"].unique()[:10])  # Print first 10 unique prices

# Step 2: Clean 'item_price' column
chipo["item_price"] = chipo["item_price"].astype(str).str.replace("$", "", regex=True).str.strip()
chipo["item_price"] = pd.to_numeric(chipo["item_price"], errors="coerce")  # Convert to float

# Step 3: Check for NaN values after conversion
print("\nRows with NaN values in item_price (if any):")
print(chipo[chipo["item_price"].isna()])

# Step 4: Filter products that cost more than $10
expensive_items = chipo[chipo["item_price"] > 10]

# Step 5: Count unique products
num_expensive_products = expensive_items["item_name"].nunique()

print(f"\nNumber of unique products that cost more than $10.00: {num_expensive_products}")

```

### Step 6. Sort by the name of the item

### Step 7. What was the quantity of the most expensive item ordered?

```python


import pandas as pd

# Load dataset
url = "https://raw.githubusercontent.com/justmarkham/DAT8/master/data/chipotle.tsv"
chipo = pd.read_csv(url, sep="\t")

# Inspect the first few rows
print(chipo.head())

# Check for missing values in all columns
print("\nMissing values in dataset:")
print(chipo.isna().sum())

# Convert 'item_price' to numeric
chipo["item_price"] = chipo["item_price"].astype(str).str.replace("$", "", regex=True).str.strip()
chipo["item_price"] = pd.to_numeric(chipo["item_price"], errors="coerce")  # Convert to float

# Check if conversion was successful
print("\nMissing values after conversion:")
print(chipo["item_price"].isna().sum())

# Drop rows with NaN values in 'item_price'
chipo = chipo.dropna(subset=["item_price"])

# Verify cleaned data
print("\nData after cleaning:")
print(chipo.head())

# Now, find the most expensive item ordered
if not chipo.empty:  # Ensure there is data
    most_expensive_idx = chipo["item_price"].idxmax()
    most_expensive = chipo.loc[most_expensive_idx]
    most_expensive_quantity = most_expensive["quantity"]
    print(f"\nThe quantity of the most expensive item ordered: {most_expensive_quantity}")
else:
    print("\nDataset is empty after cleaning. Please check the data source.")
```

### Step 8. How many times was a Veggie Salad Bowl ordered?

```python
veggie_salad_bowl_orders = chipo[chipo["item_name"] == "Veggie Salad Bowl"]["quantity"].sum()

print(f"The Veggie Salad Bowl was ordered {veggie_salad_bowl_orders} times.")
```

### Step 9. How many times did someone order more than one Canned Soda?

```python

canned_soda_orders = chipo[(chipo["item_name"] == "Canned Soda") & (chipo["quantity"] > 1)].shape[0]

print(f"Canned Soda was ordered more than once {canned_soda_orders} times.")
```
