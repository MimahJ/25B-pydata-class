#### 16. How to add a border (filled with 0's) around an existing array? (★☆☆)

#### 19. Create a 8x8 matrix and fill it with a checkerboard pattern (★☆☆)

````python
checkerboard = np.zeros((8, 8), dtype=int)
checkerboard[1::2, ::2] = 1  # Odd rows, even columns
checkerboard[::2, 1::2] = 1  # Even rows, odd columns

print (checkerboard)
````

#### 22. Normalize a 5x5 random matrix (★☆☆)

````python
matrix = np.random.rand(5, 5)  
matrix_min = matrix.min()
matrix_max = matrix.max()
normalized_matrix = (matrix - matrix_min) / (matrix_max - matrix_min)

print("Original Matrix:\n", matrix)
print("\nNormalized Matrix:\n", normalized_matrix)
````

#### 23. Create a custom dtype that describes a color as four unsigned bytes (RGBA) (★☆☆)

````python
rgba_dtype = np.dtype([
    ('r', np.uint8),  # Red channel (0-255)
    ('g', np.uint8),  # Green channel (0-255)
    ('b', np.uint8),  # Blue channel (0-255)
    ('a', np.uint8)   # Alpha channel (0-255, transparency)
])

colors = np.array([
    (255, 0, 0, 255),   # Red
    (0, 255, 0, 255),   # Green
    (0, 0, 255, 255),   # Blue
    (255, 255, 0, 128)  # Yellow (semi-transparent)
], dtype=rgba_dtype)

print(colors)
````

#### 24. Multiply a 5x3 matrix by a 3x2 matrix (real matrix product) (★☆☆)

````python
A = np.random.randint(1, 10, (5, 3))  

B = np.random.randint(1, 10, (3, 2))

C = np.dot(A, B)  # OR use C = A @ B

print("Matrix A (5x3):\n", A)
print("\nMatrix B (3x2):\n", B)
print("\nResulting Matrix C (5x2):\n", C)
````

#### 25. Given a 1D array, negate all elements which are between 3 and 8, in place. (★☆☆)

````python
arr = np.array([1, 3, 5, 7, 9, 2, 6, 8, 10])

arr[(arr > 3) & (arr < 8)] *= -1

print(arr)
````

#### 29. How to round away from zero a float array ? (★☆☆)

````python

Z = np.array([-1.7, -1.5, -0.2, 0.2, 1.5, 1.7])

rounded_Z = np.copysign(np.ceil(np.abs(Z)), Z)
print(rounded_Z)
````

#### 30. How to find common values between two arrays? (★☆☆)

````python


A = np.array([1, 2, 3, 4, 5])
B = np.array([4, 5, 6, 7, 8])

common_values = np.intersect1d(A, B)
print(common_values)
````
